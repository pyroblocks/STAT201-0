## 1) Ted Talks (45 points)

Run the code below to read the **TED_Talks.json** file you started exploring in the previous homework assignment. Remember that for the code to work properly, **the json file should be in the same folder as this ipynb file**.

```python
import json
with open("TED_Talks.json", "r") as file:
    TED_Talks_data=json.load(file)
```

### a)
**Print** the type of the TED_Talks_data.

Then, find the **headline**, **speaker** and **year_filmed** of the talk with the highest number of **views**. **Be careful with the datatypes!** **(15 points)**

```python
print(type(TED_Talks_data))

views_dict = {TED_Talks_data.index(x): int(x['views']) for x in TED_Talks_data} # create a dictionary of list index: views for all talks in TED_Talks_data
highest_views = max(views_dict, key=views_dict.get) # get the index of the talk in TED_Talks_data with the most views

print(f'The talk with the most number of views is entitled "{TED_Talks_data[highest_views]["headline"]}" and was given by {TED_Talks_data[highest_views]["speaker"]} in {TED_Talks_data[highest_views]["year_filmed"]}. It has {int(TED_Talks_data[highest_views]["views"]):,} views.')
```

### b)

Find out if the majority of talks (more than 50%) have less **views** than the overall average. You need two basic statistical tools to answer this question. You can implement them yourself or import them from a library. **(7.5 points)**

```python
import numpy as np

talk_views = list(views_dict.values())

mean_views = np.mean(talk_views)

share_less_average = len([i for i in talk_views if i < mean_views])/len(talk_views) # could also calculate this by seeing if the median is less than the mean, but this allows for less calculation in the next part

if share_less_average > 0.5:
    print(f'The majority of talks ({share_less_average:.2%}) have fewer views than the overall average.')
else:
    print(f'A minority of talks ({share_less_average:.2%}) have fewer views than the overall average.')
```

### c)

Find out if 25% or more of the talks have more **views** than the overall average. You can again implement the necessary statistical tools yourself or use a library. **(7.5 points)**

```python
share_more_average = 1-share_less_average
if share_more_average >= 0.25:
    print(f'More than 25% of talks ({share_more_average:.2%}) have more views than the overall average.')
else:
    print(f'Less than 25% of talks ({share_more_average:.2%}) have more views than the overall average.')
```

### d)

Find the **headline** of the talk that received the highest number of **Confusing** votes. **(15 points)**

```python
confusing_dict = {}

# the votes are stored under 'rates' for each talk, which itself contains several different ratings, each in a different order. So, we must iterate through each talk's 'rates' header and then go through each rating until we find the number of confusing votes
for index,talk in enumerate(TED_Talks_data): # enumerate with index so we can add it to confusing_dict later
    talk_rates = talk['rates']
    for rating in talk_rates:
        if rating['name'] == 'Confusing':
            confusing_dict[index] = int(rating['count']) # if we're at the confusing rating, create a (TED_Talks_data index: Confusing rating count) pair and add this pair to confusing_dict
            break

most_confusing = max(confusing_dict, key=confusing_dict.get) # get the index of the talk in TED_Talks_data with the most confusing ratings
print(f'The talk with the most number of Confusing votes is entitled "{TED_Talks_data[most_confusing]["headline"]}" and has {confusing_dict[most_confusing]:,} Confusing votes.')
```

### a)

Create a new class called **Text_Manager**. This class will be used to create objects that store and manage the messages in a cell phone.

- While creating an object, the class should take one input, which will be a list of dictionaries like the inbox given in the previous cell.
- The class should save its input as **its only attribute**. Call this attribute **messages**. Note that this attribute will store all the messages in the inbox. 
- After the attribute, the class should have the following five methods:
    - **add_new_message**: This method takes the information of a new message as **four inputs**: from_number, time_arrived, date_arrived, and text. It adds the new message as a new dictionary to the inbox. It sets the **has_been_viewed** status for the new message as **False**.
    - **message_count**: This method does not take any inputs and returns the number of all messages in the inbox.
    - **get_unread_messages:** This method does not take any inputs and returns the information of the unread messages in the inbox as the output. All the information except **has_been_viewed** should be returned for each unread message. Lastly, the **has_been_viewed** status of the returned messages should change from False to True.
    - **delete_message:** This method takes the index of a message as the only input and deletes the corresponding message in the inbox.
    - **clear_inbox:** This method takes no inputs and deletes all messages in the inbox.
 
**(38 points)**

```python
class Text_Manager:
    def __init__(self, dict_list):
        self.messages = dict_list

    def add_new_message(self, from_number, time_arrived, date_arrived, text):
        self.messages.append({'has_been_viewed': False, 'from_number': from_number, 'time': time_arrived, 'date': date_arrived, 'text': text})

    def message_count(self):
        return(len(self.messages))

    def get_unread_messages(self):
        messages_to_return = []
        for index, message in enumerate(self.messages): # enumerate with index so that we can edit self.messages to mark the message as viewed
            if message['has_been_viewed'] == False:
                messages_to_return.append(f'You have an unread message sent by {message["from_number"]} on {message["date"]} at {message["time"]}. It reads: {message["text"]}')
                self.messages[index]['has_been_viewed'] = True
        return(messages_to_return)

    def delete_message(self, message_index):
        self.messages.pop(message_index)

    def clear_inbox(self): # could use delete_message to delete every message one-by-one, but the end result we want is an empty list, so we can just set self.messages to that to begin with
        self.messages = []
```
