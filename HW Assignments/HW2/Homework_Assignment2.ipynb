In this question, you will estimate pi using an iterative method. Pi can be computed using the Gregory-Leibniz Series as:

$$\pi = 4 - \frac{4}{3} + \frac{4}{5} - \frac{4}{7} + \frac{4}{9} - ... = \sum\limits_{n=1}^{\infty} (-1)^{k+1}\frac{4}{2k-1}$$

Keep adding terms until your subsequent estimates are within 0.0001 of each other. At the end, display the pi estimate to **five decimal points** and the number of terms you added. **(25 points)**

**Hints**: 
1. Note that before the loop, you need to initialize three different variables: 
    - A **sum_new** variable to keep adding the new term
    - A **sum_old** variable to store the sum of the previous iteration before adding the new term and compare the difference at each iteration
    - A **k** variable to calculate the new term and also to keep track of the number of terms.
2. There should be a **loop** that checks if the difference between **sum_new** and **sum_old** is larger than 0.0001 and if it is, updates **sum_old**, updates **sum_new** and increments **k**.
3. At the end, display your pi estimate and the number of total terms using formatted text and the correct variables.
4. To limit your output to five decimal points, you need to put something inside the curly braces - you need to look that up.

```python
sum_new = 4-(4/3)
sum_old = 4
k = 2
while abs(sum_new-sum_old) > 0.0001:
    k += 1
    sum_old = sum_new
    sum_new = sum_new + (-1)**(k+1) * (4/(2*k-1))
print(f'Pi estimate: {sum_new:.5f} achieved with {k} total terms')
```

### 4)

Write a **function** called **number_of_factors** that takes a positive integer as the only input and returns how many factors it has as the only output. Call the function with 190 to check if it works. You do not need to check for invalid inputs.

**Please note that you need to use a user-defined function with its inputs and outputs. Any code that uses an input line will not receive any credit.**

**(7.5 points)**

```python
def number_of_factors(pos_int):
    counter = 1 # start at 1 to include the number itself
    for i in range(1,int(pos_int/2)+1): # we know that all numbers greater than half of the integer besides the integer itself will not be factors, so we can skip those
        if pos_int % i == 0:
            counter += 1
    return counter

number_of_factors(190)
```

### 6)

Write a **function** called **area_rect** that calculates the area of a rectangle. It should take the length (in inches) and the width (in inches) as two numeric inputs and return the area in square inches and the area in square feet as two outputs. There are 12 inches in 1 foot.

When the function is called, it should also:
- Print "The area in square inches is \_\_\_." and "The area in square feet is \_\_\_." in two separate lines, where \_\_\_ stands for the first and the second output, respectively.
- Be able to take only one input, in which case it should do all of the above for a square.

Call the function with 102 and 60 and then only with 60 to check if it works. You do not need to check for invalid inputs.

**Please note that you need to use a user-defined function with its inputs and outputs. Any code that uses the input function will not receive any credit.**

**(15 points)** 

```python
def area_rect(length_in,width_in=None):
    if width_in == None:
        width_in = length_in
    area_sqin = length_in*width_in
    area_sqft = area_sqin/144

    print("The area in square inches is {}. The area in square feet is {}.".format(area_sqin, area_sqft))
    return(area_sqin, area_sqft)

print(area_rect(102,60))
print(area_rect(60))
```
