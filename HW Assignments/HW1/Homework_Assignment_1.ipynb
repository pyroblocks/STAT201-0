### 5)

Write a program that asks the user to enter either **rock**, **paper** and **scissors**. The program needs to determine and print if the user wins, loses, or ties the computer at the game of Rock, Paper, Scissors. Keep in mind that rock beats scissors; scissors beat paper; paper beats rock.

Your program also needs to:

- Print a meaningful sentence that includes the winner (or states that it is a tie), the computer's choice and the user's choice.
- Handle case sensitivity. For example, if the user enters **Rock**, instead of **rock**, the code should still run properly.
- Print **"Invalid choice!"**, if the user enters a word other than one of the three valid choices.

The first two lines of the code, which create the computer's choice, are given below. Your code should come below them. Run the program with a valid input of your choice.

**(20 points)**

```python
# starter code to generate a random choice of rock, paper, scissors
import random as rm
comp_choice = rm.choice(['rock', 'paper', 'scissors'])

# Your code here
user_choice = input("Please pick rock, paper, or scissors.").lower()

# We re-use the winner script a lot, so we can save having to rewrite it for each use case here
user_wins = "The player has won! ({} beats {})"
comp_wins = "The computer has won! ({} beats {})"
tie = "It's a tie! (both chose {})"

if user_choice == 'rock':
    if comp_choice == 'rock':
        print(tie.format(user_choice))
    elif comp_choice == 'scissors':
        print(user_wins.format(user_choice, comp_choice))
    elif comp_choice == 'paper':
        print(comp_wins.format(comp_choice, user_choice))

elif user_choice == 'paper':
    if comp_choice == 'paper':
        print(tie.format(user_choice))
    elif comp_choice == 'rock':
        print(user_wins.format(user_choice, comp_choice))
    elif comp_choice == 'scissors':
        print(comp_wins.format(comp_choice, user_choice))

elif user_choice == 'scissors':
    if comp_choice == 'scissors':
        print(tie.format(user_choice))
    elif comp_choice == 'paper':
        print(user_wins.format(user_choice, comp_choice))
    elif comp_choice == 'rock':
        print(comp_wins.format(comp_choice, user_choice))

else:
    print('Invalid choice!')
```
