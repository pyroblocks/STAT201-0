## 1) Lists

### a) Creating and indexing 

```python
# A Python list:
# Everything between []
# All elements separated by ,
# ordered - elements have positional indices
# mutable - you can alter the data (add, take out etc.)

roster = ['Jaylen Brown', 'Jayson Tatum', 'Jrue Holiday', 'Kristaps Porzingis', 'Derrick White', 'Al Horford',
'Payton Pritchard', 'Jaden Springer', 'Baylor Scheierman',
'Xavier Tillman', 'Neemias Queta', 'Sam Hauser',
'Luke Kornet', 'Jordan Walsh']

print(type(roster))

# Indexing an element
print(roster[0]) # first element
print(roster[2])  # third element

# slicing a range of elements
print(roster[1:4]) # second to tenth player roster[1], roster[2], roster[3]

print(roster[2:10:2]) # third to tenth player with a stepsize of 2 (2,4,6)

print(roster[-1]) # last player

print(roster[-1:-5:-1]) # If you want to slice backwards, you need negative stepsize (end index excluded again)

print(roster[:5]) # everything until index 5( excluded)

print(roster[7:]) # everything from index 7( included)

# Using indices we can alter the data
roster[2] = 'P_3' # this will replace the third element value with 'P_3' value
print(roster)

roster[2:4] = ['P 3', 'P 4'] # Careful about the lengths matching
print(roster)

# Lastly, you can return the index of a given element by .index() method
roster.index('Derrick White') # .index is a list method
```

### b) Handling Multiple Lists - Concatenating

```python
# important note: A list can have different data type

list1 = ['A', 1, 3.4, True] # Strings, floats, integers and boolean all together
list2 = ['B', False, 3,7]
list3 = [2,3,4,5,6]

# Putting all the data together
nested_list = [list1, list2, list3, False, 5, 'C'] # This is a list of lists
print(nested_list)

print(nested_list[0])

# How to access the data inside the nested lists - use two (or more) sets of indices
print(nested_list[0][2])
print(nested_list[3])
# Same idea applies to slicing


# You can also concatenate multiple lists - using +
print(list1+list2+list3)

print(list1+ nested_list)
```

### e) Unpacking a Tuple

```python
my_tuple = (1, 'a', 3.5)

#unpack a tuple

x,y,z = my_tuple
print(x)
print(y)
print(z)

roster_tuple = ('Jaylen Brown', 'Jayson Tatum', 'Jrue Holiday', 'Kristaps Porzingis', 'Derrick White', 'Al Horford',
'Payton Pritchard', 'Jaden Springer', 'Baylor Scheierman',
'Xavier Tillman', 'Neemias Queta', 'Sam Hauser',
'Luke Kornet', 'Jordan Walsh')

n1,*_ ,n_l = roster_tuple # This line unpacks the first, last element
print(n1)
print(_)
print(n_l)

n0,n1,*_ ,n_l = roster_tuple
print(n1)

_, n1, *_, n_l = roster_tuple # This line unpacks the second, and last element

print(n1)
print(n_l)
print(_)

*_, n1, _, n_l = roster_tuple # This line unpacks the last and the third last element

print(n1)
print(n_l)
print(_)

*start, last = roster_tuple
print(start)
print(last)
print('-------------')
start, *last = roster_tuple
print(start)
print(last)
```

### f) Iterating - Introducing a List of Tuples

```python
#it is possible iterate through elements in parallel
    # using multiple iterators

list_of_tuple = [(1,100),(2,99),(3,98)]

for elem in list_of_tuple:
    print(elem)

# While iterating through the list, I can use two iterators for the two numbers in each tuple

for num1,num2 in list_of_tuple: # num1 becomes the first number in each tuple, num2 becomes the second
    print(num1)
    print(num2)
    print(num1,num2)

# Tuples need to be at the same length
# this is also possible with a tuple of tuples
```

### g) Advanced Example with zip

```python
roster = ['Jaylen Brown', 'Jayson Tatum', 'Jrue Holiday', 'Kristaps Porzingis', 'Derrick White', 'Al Horford',
'Payton Pritchard', 'Jaden Springer', 'Baylor Scheierman',
'Xavier Tillman', 'Neemias Queta', 'Sam Hauser',
'Luke Kornet', 'Jordan Walsh']

jersey_nums = [7, 0, 4, 8, 9, 42, 11, 44, 55, 26, 88, 30, 40, 27]
# One thing to know: The name and the jersey number in the same index belong to the same player.

pair = list(zip(roster,jersey_nums))# Converting the zip output to a list/tuple right away makes it permanent
                                        # otherwise it is for one-time use but takes much less space in memory

print(pair)

# If the lengths are not matching, the items that don't match gets skipped
# You can bring more than two lists/tuples together with zip
# Works with tuples as well

for name, num in pair:
    print(name, num)
```

## 4) Dictionaries

```python
# A dict has a different structure than lists/tuples.

# They are mutable - you can change the content
# The idea of being ordered is different for a dict

# The elements of a dict consist of key-value pairs:
    # Values are similar to the elements of lists/tuples
    # Keys are like tags for the values - they define/explain them.
```

### a) Creating and Indexing (with keys)

```python
# A dict is created with {}
# Each element is written as key:value

brown = {'pos':'SG', 'age':27, 'height':78, 'weight':220, 'college':'UC Berkeley', 'salary':49.2}

# Indexing using keys

print(brown['age'])
print(brown['salary'])

# There are no positional indices
# print(brown[1]) -- > error

# brown['age': 'weight'] ---> error, slicing is not allowed

# To add a new element (key-value pair), you need to use the key, as if it already exists in the dict
brown['new_key'] = 55 # "new_key":55 is added to the dictionary

print(brown)

# Change the value of an element
brown['height'] = 79# Accessing with the key and assigning the value to be overwritten
print(brown)

dict_keys = list(brown.keys())
print(dict_keys)# Converting all keys to a list, can change to tuple too

# Return all the values
dict_values = list(brown.values())
print(dict_values)

dict_items = list(brown.items())# This is a list of tuples (can be iterated with multiple iterators)
print(dict_items)

print(dict_items[1][1])
```

```python
# toy examples

toy_exam = {1:"a", 2.5:'xyz', "age":25} # keys can be numbers too
toy_exam[1] # This works because 1 is a key, not a positional index

# keys needs to be unique
toy_exam1 = {1:"a", 2.5:'xyz', "age":25, 2.5:[45,2]} # Does not throw an error 
# ignores all the elements with the same key except the last one
print(toy_exam1)
```

### b) Handling Multiple Dictionaries - Concatenating

```python
# There is a way to concat dicts (not with the + sign) but it is both out of scope and counter-intuitive

# The common practice is to nest multiple dicts in a main structure

brown = {'pos':'SG', 'age':27, 'height':78, 'weight':220, 'college':'UC Berkeley', 'salary':49.2}
tatum = {'pos':'SF', 'age':26, 'height':80, 'weight':210, 'college':'Duke', 'salary':34.8}

roster_details = [brown, tatum]
print(roster_details)

print(roster_details[1])
print(roster_details[1]['weight'])
```

```python
# You can also nest a dict inside another, but ONLY AS A VALUE
tatum = {'pos':'SF', 'age':26, 'height':80, 'weight':210,'college':'Duke', 'salary':34.8, 'fav_teammate':brown}
print(tatum)

print(tatum['fav_teammate'])
print(tatum['fav_teammate']['weight'])# Using multiple keys to access the nested dict

# The values of a dict can be anything: a list, tuple, another dict, int, string etc.
# The keys can be int,float,str,boolean (also a tuple)
```

### c) Adding and Removing Elements

```python
brown = {'pos':'SG', 'age':27, 'height':78, 'weight':220, 'college':'UC Berkeley', 'salary':49.2}

# To add a new element (key-value pair), you need to use the key, as if it already exists in the dict
brown['new_key'] =55

print(brown)

# adding multiple elements with .update() - input should be another dict
brown.update({'ab':[1,2,3], 4.5:'abc'})
print(brown)
```

```python
# To delete an element (key-value pair), use .pop() with the key as the input

brown.pop('age')
print(brown)

# careful: if you have run this cell, and you run it again it will cause an error
#The first time you run it, it will delete champ, thus the second time you run it, the dict brown won't have champ to be deleted
```

### d) Iterating

```python
# Let's create a dict from two list
roster = ['Jaylen Brown', 'Jayson Tatum', 'Jrue Holiday', 'Kristaps Porzingis', 'Derrick White', 'Al Horford',
'Payton Pritchard', 'Jaden Springer', 'Baylor Scheierman',
'Xavier Tillman', 'Neemias Queta', 'Sam Hauser',
'Luke Kornet', 'Jordan Walsh']

jersey_nums = [7, 0, 4, 8, 9, 42, 11, 44, 55, 26, 88, 30, 40, 27]

# create a dict from two lists

# start with an empty dict
roster_dict = {}

for i in range(len(roster)): # iter through the index range of the lists
    roster_dict[roster[i]] = jersey_nums[i]  #make the ith name the key and the ith num the value

print(roster_dict)
```

```python
roster_dict = {}

# Here make the ith num the key and the ith name the value
for i in range(len(roster)):
    roster_dict[jersey_nums[i]] = roster[i]

print(roster_dict)
```

```python
# Iterating through a dict itself

# through the keys
for nums in roster_dict.keys():
    print(nums)

# through the values
for vals in roster_dict.values():
    print(vals)


# Iterating through the key and the val in parallel (using multiple iterators)
for x,y in roster_dict.items():
    print(x,y)

for elem in roster_dict.items():
    print(elem)
```

### e) Using as Function Input

```python
brown = {'pos':'SG', 'age':27, 'height':78, 'weight':220, 'college':'UC Berkeley', 'salary':49.2}
# max() and min() when applied directly to a dictionary
# operate only on its KEYS (not values)
# so this returns the alphabetically largest/smallest key
max(brown)   #  'weight' (alphabetically last key)
min(brown)   # 'age' (alphabetically first key)

# Note for a dictionary like: dict_eg = {"a":5, "b":"C", 5:"D"}, max(), min() will give error. 
# Cant't compare string with numeric keys

# len() returns the number of key–value pairs in the dictionary
len(brown)   

# sum(brown) would cause an error
```
